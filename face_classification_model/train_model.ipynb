## What Does `mltk train train_model.py` Do?

When you run this command, MLTK will:

1. **Load Configuration** - Read your `train_model.py` file to get model architecture and training settings (epochs=50, batch_size=64, learning rate, etc.)

2. **Load Dataset** - Load training and validation data from `./dataset/dataset` folder with your 2 classes: `face` and `noface`

3. **Build Model** - Create the MobileNet-based neural network:
   - Input shape: 112x112 grayscale images
   - Base model: MobileNet with alpha=0.25 (lightweight)
   - Output: 2 classes (face/noface) with softmax activation

4. **Preprocess Images** - Apply preprocessing to images:
   - Data augmentation (rotation, shifting, zooming, flipping)
   - Normalization: `(img / 127.5) - 1.0`

5. **Train Model** - Train for 50 epochs with:
   - Batch size: 64
   - Optimizer: Adam
   - Loss function: Categorical crossentropy
   - Monitoring: Validation accuracy
   - Early stopping if no improvement for 12 epochs
   - Learning rate reduction if loss plateaus

6. **Generate Output** - Save trained model files:
   - TensorFlow model format
   - TFLite model (optimized for embedded devices like your microcontroller)
   - Model archive with metadata

7. **Evaluation** (Optional) - You can add evaluation and profiling after training completes
